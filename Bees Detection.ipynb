```python
import numpy as np
import cv2 as cv
```

```python
video=cv.VideoCapture('bee1short.mp4')
```

```python
fgbg= cv.createBackgroundSubtractorKNN(1000,400,False)
```

```python
video=cv.VideoCapture('bee1short.mp4')
i=0
while True:
    ret, frame = video.read()
    if frame is None:
        print('Cannot Find Video: Check Path')
        break
    if (i>=0): 
        #applying background subtraction
        fgmask = fgbg.apply(frame)
        #fgmask= cv.GaussianBlur(fgmask, (5, 5), 0)

        #finding area after removing background
        contours, _= cv.findContours(fgmask, cv.RETR_LIST, cv.CHAIN_APPROX_NONE)
        totalarea=0.0
        for contour in contours:
          area = cv.contourArea(contour)
          if area>1500:
            cv.drawContours(frame, contour, -1, (250, 0, 0), 3)
            totalarea+=area
        BeeCount= totalarea/14000
        #print(frame)
        print(totalarea)

        #shows frame number in original video
        cv.rectangle(frame, (10, 2), (200,20), (255,255,255), -1)
        #cv.putText(frame, "Frame:"+str(video.get(cv.CAP_PROP_POS_FRAMES)),(15, 15),cv.FONT_HERSHEY_SIMPLEX,0.5,(0,0,0))
        cv.putText(frame, "Area:"+str(totalarea),(15, 15),cv.FONT_HERSHEY_SIMPLEX,0.5,(0,0,0))

        cv.namedWindow('frame', cv.WINDOW_KEEPRATIO)
        cv.imshow('frame',frame)
        cv.resizeWindow('frame', 1000, 1600)
        #cv.imshow('fgmask',fgmask)
        
        cv.imwrite("image"+str(i)+".jpg",frame)
        print("image"+str(i)+".jpg stored")
        #cv.imwrite("image2.jpg",fgmask)

        #press key q to 
        keyboard = cv.waitKey(30)
        if keyboard == 'q' or keyboard == 27:
            break
    i=i+1
video.release()
cv.destroyAllWindows()
```
